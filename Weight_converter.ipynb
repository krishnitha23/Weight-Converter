# ***WEIGHT*** ***CONVERTER***

```python
weight = float(input("ENTER YOUR WEIGHT: "))
```

```python
unit = input("KILOGRAMS OR POUNDS?(K OR L): ")
```

```python
if unit == "K":
   weight = weight*2.205
   unit = "kgs."
   print(f"YOUR WEIGHT IS : {round(weight,1)} {unit}")
elif unit == "L":
   weight = weight/2.205
   unit = "lbs."
   print(f"YOUR WEIGHT IS : {round(weight,1)} {unit}")
else:
   print(f"OOPS! {unit} unit was not valid!")
```
